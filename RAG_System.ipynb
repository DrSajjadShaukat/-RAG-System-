### What is RAG?
RAG = Retrieval Augmented Generation
Instead of relying only on AI training data,
RAG lets you add YOUR OWN documents and
get answers from them!

- Custom knowledge base
- Intelligent document search
- Accurate answers from your data

```python
import warnings
warnings.filterwarnings('ignore')
import os
import anthropic
from google.colab import userdata

# Setup
client = anthropic.Anthropic(
    api_key=userdata.get("ANTHROPIC_API_KEY")
)

# ============================================
# KNOWLEDGE BASE — Add your documents here!
# ============================================
knowledge_base = [
    "Python is a programming language great for AI development.",
    "LangChain is a framework for building AI agent applications.",
    "RAG stands for Retrieval Augmented Generation.",
    "ChromaDB is a vector database for storing embeddings.",
    "Dr. Sajjad Shaukat is a Math PhD learning AI Engineering in Houston.",
    "Texas Southern University is located in Houston Texas.",
    "Anthropic created Claude AI which is used for natural language processing.",
    "Embeddings convert text into numbers that capture meaning and context.",
    "Vector databases store embeddings for fast similarity search.",
    "LLMs are Large Language Models trained on massive text datasets.",
]

# ============================================
# RAG FUNCTIONS
# ============================================

def search_knowledge(question):
    """Find relevant documents from knowledge base"""
    question_words = question.lower().split()
    relevant = []

    for doc in knowledge_base:
        doc_lower = doc.lower()
        matches = sum(1 for word in question_words
                     if word in doc_lower and len(word) > 3)
        if matches > 0:
            relevant.append((matches, doc))

    relevant.sort(reverse=True)
    return [doc for _, doc in relevant[:3]]

def rag_agent(question):
    """Answer questions using knowledge base"""
    print(f"Question: {question}")
    print("Searching knowledge base...")

    relevant_docs = search_knowledge(question)

    if relevant_docs:
        context = "\n".join(relevant_docs)
        print(f"Found {len(relevant_docs)} relevant documents!")
    else:
        context = "No specific information found in knowledge base."
        print("No relevant documents found.")

    response = client.messages.create(
        model="claude-haiku-4-5",
        max_tokens=1024,
        system="You are a helpful assistant. Answer questions using only the provided context. If context doesn't have the answer, say so honestly. Use plain text only.",
        messages=[{
            "role": "user",
            "content": f"Context:\n{context}\n\nQuestion: {question}"
        }]
    )
    return response.content[0].text

# ============================================
# ADD YOUR OWN DOCUMENTS
# ============================================
def add_document(text):
    """Add new document to knowledge base"""
    knowledge_base.append(text)
    print(f"Added: {text[:50]}...")

# ============================================
# TEST THE RAG SYSTEM
# ============================================
print("=" * 50)
print("RAG System Ready!")
print(f"Knowledge base: {len(knowledge_base)} documents")
print("=" * 50)

print("\nTest 1:")
print(rag_agent("What is RAG?"))

print("\nTest 2:")
print(rag_agent("Who is Dr. Sajjad?"))

print("\nTest 3:")
print(rag_agent("What is LangChain used for?"))

print("\nTest 4 — Adding new document:")
add_document("Cryptography is the study of secure communication techniques.")
print(rag_agent("What is cryptography?"))
```
